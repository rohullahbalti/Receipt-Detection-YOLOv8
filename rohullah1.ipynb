```python
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt

# Model load
model = YOLO("/content/drive/MyDrive/best2.pt")

# Image path
image_path = "/content/drive/MyDrive/sample-receipts.png"

# Predict
results = model.predict(source=image_path, conf=0.5)

# Original image open
orig_img = Image.open(image_path)

# Crop receipts
for i, box in enumerate(results[0].boxes.xyxy):
    x1, y1, x2, y2 = map(int, box.tolist())  # coordinates int me
    cropped = orig_img.crop((x1, y1, x2, y2))

    # Display
    plt.imshow(cropped)
    plt.axis("off")
    plt.title(f"Cropped Receipt {i+1}")
    plt.show()

```
